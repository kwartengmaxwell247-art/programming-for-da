```python
# 1. Create a valid CSV file with at least 3 products
csv_filename = "products_dict.csv"
csv_content = """product,quantity,price
Laptop,5,1200.00
Smartphone,12,800.50
Headphones,25,150.00
"""

with open(csv_filename, "w", newline="") as file:
    file.write(csv_content)
print(f"'{csv_filename}' successfully created!\n")


# 2, 3, & 4. Read file, display contents, and handle missing files safely
try:
    with open(csv_filename, mode="r", newline="") as file:
        # Load the CSV data using DictReader
        reader = csv.DictReader(file)
        
        # 3. Display Column Names (fieldnames contains the headers)
        print("--- Column Names ---")
        print(reader.fieldnames)
        
        # 3. Display the Rows
        print("\n--- Rows (Dataset Content) ---")
        for row in reader:
            # Each 'row' is an ordered dictionary
            print(dict(row))

except FileNotFoundError:
    print(f"⚠️ Error: The file '{csv_filename}' could not be found.")
```

```python
print("=== TASK 1: Two 3x3 Matrices ===")
# Arrays are filled with non-zero numbers so division works perfectly
A = np.array([[1, 2, 3],
              [7, 8, 9]])

B = np.array([[9, 8, 7],
              [3, 2, 1]])
print("Matrix A:\n", A)
print("\nMatrix B:\n", B)


print("\n=== TASK 2: Basic Operations ===")
print("Addition (A + B):\n", A + B)
print("\nSubtraction (A - B):\n", A - B)
print("\nElement-wise Multiplication (A * B):\n", A * B)
print("\nDivision (A / B):\n", A / B)


print("\n=== TASK 3: Aggregations ===")
print("Grand Total of Matrix A:", A.sum())
print("Row Sums of A (axis=1):", A.sum(axis=1))
print("Column Means of A (axis=0):", A.mean(axis=0))


print("\n=== TASK 4: Multiplication Types ===")
print("Element-wise (A * B):\n", A * B)
print("\nMatrix Multiplication (A @ B):\n", A @ B)
```

```python
import numpy as np
print("=== TASK 1: Create Product and Revenue Arrays ===")
# Create parallel arrays matching 4 products to their revenues
products = np.array(["Headphones", "Laptops", "Smartphones", "Tablets"])
revenue = np.array([1500, 12000, 8500, 3400])

print("Products:", products)
print("Revenues:", revenue)


print("\n=== TASK 2: Sort Revenue ===")
# Sort ascending (lowest to highest)
sorted_asc = np.sort(revenue)
# Sort descending (highest to lowest) by reversing the sorted array
sorted_desc = sorted_asc[::-1]

print("Revenue (Ascending):", sorted_asc)
print("Revenue (Descending):", sorted_desc)


print("\n=== TASK 3: Use argsort() to Rank Products ===")
# np.argsort returns the indices that would sort the array ascending
sorted_indices = np.argsort(revenue)
# Reverse indices to get descending order (highest revenue first)
leaderboard_indices = sorted_indices[::-1]

print("Indices that sort revenue (Descending):", leaderboard_indices)
print("\n--- Final Paired Leaderboard ---")
for rank, idx in enumerate(leaderboard_indices, start=1):
    print(f"Rank {rank}: {products[idx]} - ${revenue[idx]}")


print("\n=== TASK 4: Report the Top Two Products ===")
# Grab the first two items using slice [:2] from our ranked leaderboard indices
top_two_indices = leaderboard_indices[:2]

print("🏆 TOP PRODUCTS LEADERBOARD 🏆")
for rank, idx in enumerate(top_two_indices, start=1):
    print(f"🥇 No.{rank}: {products[idx]} (Revenue: ${revenue[idx]})")
```
